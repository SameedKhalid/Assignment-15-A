```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('ride_hailing_dataset.csv')
print(df.shape)
```

### Q1: What is the average fare by service type?

```python
print(df.groupby('service_type')['fare'].mean().round(2).sort_values(ascending=False))
```

**Answer:** **InDrive** has the highest average fare (PKR 573.51) and **Uber** the lowest (PKR 486.13).

### Q2: Which city has the most trips?

```python
print(df['city'].value_counts())
```

**Answer:** **Islamabad** has the most trips (29).

### Q3: What is the average distance per trip?

```python
print(round(df['distance_km'].mean(), 2))
```

**Answer:** The average distance is **9.61 km** per trip.

### Q4: Compare trip ratings by gender

```python
rg = df.groupby('gender')['trip_rating'].agg(['mean','count']).round(2)
print(rg)
sns.barplot(x=rg.index, y=rg['mean'], hue=rg.index, legend=False)
plt.ylim(0, 5)
plt.title('Average Trip Rating by Gender')
plt.show()
```

**Answer:** Female: 2.82; Male: 3.33. Male riders give higher ratings on average.

### Q5: Bar chart of service type vs. average fare

```python
avg = df.groupby('service_type')['fare'].mean().sort_values(ascending=False)
sns.barplot(x=avg.index, y=avg.values, hue=avg.index, legend=False)
plt.title('Average Fare by Service Type')
plt.xlabel('Service Type'); plt.ylabel('Average Fare (PKR)')
plt.show()
```

**Answer:** The chart shows InDrive as the most expensive and Uber as the cheapest service on average.
